```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Load the CSV file
# Replace 'your_file.csv' with the path to your actual CSV file
data = pd.read_csv('largecv_energy_biased_dis.csv')
```

```python
# Step 2: Extract the first column (assuming it's unnamed or named)
first_column = data.iloc[:, 0]  # this grabs the first column

# Step 3: Plot a histogram
plt.figure(figsize=(10, 6))
plt.hist(first_column, bins=200, color='skyblue', edgecolor='black')  # Adjust bins as needed
plt.title('Distribution of the CVs in Umbrella Sampling')
plt.xlabel('Value')
plt.ylabel('Frequency')

# Step 4: Show the plot
plt.show()
```
